# Netflix Data Analysis

This project analyzes Netflix movies and TV shows dataset to discover trends in content type, genres, countries, and release patterns using Python.

Tools Used:
- Python
- Pandas
- Matplotlib
- Seaborn

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("netflix_titles.csv")
df.head()
```

```python
df.shape
df.info()
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
sns.countplot(x='type', data=df)
plt.title('Movies vs TV Shows on Netflix')
plt.show()
```

```python
df['country'].value_counts().head(10).plot(kind='bar')
plt.title('Top 10 Countries with Most Titles on Netflix')
plt.show()
```

```python
df['listed_in'].value_counts().head(10).plot(kind='bar')
plt.title("Most Popular Genres on Netflix")
plt.show()
```

```python
df['release_year'].value_counts().sort_index().plot()
plt.title("Netflix Content Release Over Years")
plt.xlabel("Year")
plt.ylabel("Number of Titles")
plt.show()
```

```python
df.shape
```

```python
df['listed_in'].value_counts().head(10).plot(kind='bar')
plt.title("Top 10 Genres on Netflix")
plt.xlabel("Genre")
plt.ylabel("Count")
plt.show()
```

```python
# top rating
df['rating'].value_counts().plot(kind='bar')
plt.title("Distribution of Content Ratings on Netflix")
plt.xlabel("Rating")
plt.ylabel("Count")
plt.show()
```

```python
# data cleaning
df['director'] = df['director'].fillna("Unknown")
df['cast'] = df['cast'].fillna("Unknown")
df['country'] = df['country'].fillna("Unknown")
```

```python
df.isnull().sum()
```

```python
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')
```

```python
df['date_added'].head()
```

```python
df['date_added'].dt.year.value_counts().sort_index().plot()
plt.title("Content Added to Netflix Per Year")
plt.xlabel("Year")
plt.ylabel("Number of Titles")
plt.show()
```

```python
top_directors = df['director'].value_counts().head(10)

top_directors.plot(kind='bar')
plt.title("Top 10 Directors on Netflix")
plt.xlabel("Director")
plt.ylabel("Number of Titles")
plt.show()
```

## Key Insights

From the analysis of the Netflix dataset, several trends can be observed:

1. Netflix contains significantly more **Movies than TV Shows**.

2. The **United States produces the highest number of titles** on Netflix.

3. Popular genres include **Drama, International Movies, and Comedies**.

4. The number of titles released on Netflix **increased significantly after 2015**, showing the platform's rapid expansion.

5. The most common content ratings are **TV-MA and TV-14**, indicating a large amount of mature and teen content.

6. A few directors contribute multiple titles to the platform, though most titles have unique directors.
